```python
import pandas as pd
import numpy as np
```

```python
df = pd.read_csv("employee_raw_for_cleaning_project.csv")
```

```python
df
```

```python
df.columns
```

```python
df.drop(columns=["LeaveOrNot"], inplace= True)
```

```python
df.isnull().sum() * 100/ len(df)
```

```python
df['Age'].unique()
```

```python
df['Age'] = df['Age'].fillna(0)
```

```python
df['Age'] = df['Age'].replace(0,np.nan) 
```

```python
df['Age'] = df['Age'].fillna(df['Age'].median()) 
```

```python
df
```

```python
df.isnull().sum() * 100/ len(df)
```

```python
df['PerformanceScore']
```

```python
df['PerformanceScore'] = df['PerformanceScore'].fillna(0)
```

```python
df['PerformanceScore'] = df['PerformanceScore'].replace(0,np.nan)
```

```python
df['PerformanceScore'] = df['PerformanceScore'].fillna(df['PerformanceScore'].mean())
```

```python
df.isnull().sum() * 100/len(df)
```

```python
df['Salary'] = df['Salary'].fillna(0)
```

```python
df['Salary'] = df['Salary'].replace(0,np.nan)
```

```python
df['Salary'] = df['Salary'].fillna(df['Salary'].median())
```

```python
df
```

```python
df['PerformanceScore'] = df['PerformanceScore'].round(2)
```

```python
df.to_csv("cleaned_csv_file.csv", index = False)
```
